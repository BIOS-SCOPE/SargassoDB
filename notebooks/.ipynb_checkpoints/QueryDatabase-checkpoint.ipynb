```python
from sqlalchemy.orm import Session
stmt = text('SELECT cruise_id FROM cruises WHERE cruise_id = :cruise_id')
with Session(engine) as session:
    result = session.execute(stmt,{'cruise_id': 'AE1627'})
    for row in result:
        print(f"x: {row.cruise_id}")
```

```python
from sqlalchemy.engine import reflection

# # Create connection string & engine
# connection_string = "sqlite:///my_existing_database.db"
# engine = sa.create_engine(connection_string)

# Performs database schema inspection
# inspector = reflection.Inspector.from_engine(engine) #deprecated
inspector = inspect(engine)

# Get a list of all table names
table_names = inspector.get_table_names()
print(f"Table names: {table_names}")
```

```python
# Get column information for a specific table
for column in inspector.get_columns('asv_metadatas'):
    print(column['name'], column['type'], column['nullable'])

# You can also get other info
print(inspector.get_foreign_keys('users'))
print(inspector.get_indexes('users'))
```

```python
engine = create_engine(database_url)
inspector = inspect(engine)

# Get a list of table names
table_names = inspector.get_table_names()

print("Tables in the database:")
for table_name in table_names:
    print(f"- {table_name}")
```

```python
#mmmm this gives me all rows in the table
# Assuming a 'users' table is already defined or reflected
metadata = MetaData()
users_table = db.Table('cruises', metadata, autoload_with=engine)

# Build the query: Equivalent to 'SELECT * FROM users'
query = db.select(users_table)

# Execute the query and fetch results
with engine.connect() as connection:
    result = connection.execute(query)
    # Fetch all results as a list of Row objects
    rows = result.fetchall()
    for row in rows:
        print(row)

# Filter example: Equivalent to 'SELECT * FROM users WHERE name = "Alice"'
query = db.select(users_table).where(users_table.columns.program == 'BIOSSCOPE')
with engine.connect() as connection:
    result = connection.execute(query).fetchall()
```

```python
from sqlalchemy import create_engine
from sqlalchemy.ext.automap import automap_base
from sqlalchemy.orm import Session

# 1. Create the engine (replace with your database URI)
# Example URIs: 'sqlite:///existing_db.sqlite' or 'postgresql://user:pass@host/dbname'
engine = create_engine("sqlite:///../sql_app.db")

# 2. Use automap to reflect the schema
Base = automap_base()
Base.prepare(autoload_with=engine)

# 3. Access the generated classes (tables)
# Replace 'users' with the actual name of a table in your database
# User = Base.classes.users
User = Base.classes.cruises

# 4. Create a session and use it
session = Session(engine)
```
