```python
import pandas as pd 
import os
#need this to see the full column width
pd.set_option('display.max_colwidth', None)
```

```python
dataDir = '../test_data/Luis_fileLists'
files = [f for f in os.listdir(dataDir) if f.endswith('.txt')]
```

```python
files
```

```python
df = pd.read_csv(os.path.join(dataDir,'only_in_cyverse.txt'),sep='\t',header=None,comment = '#',names = ['filename'])

#strip off the end of the filename
for index,row in df.iterrows():
    #file = os.path.basename(row.to_string()).strip('.gz')
    file = os.path.basename(row['filename'])
    df.loc[index,'filename'] = file

df['source'] = 'cyverse'
df_co = df
```

```python
df_co.head()
```

```python
df = pd.read_csv(os.path.join(dataDir,'only_in_gdrive.txt'),sep='\t',header=None,comment = '#',names = ['filename'])
#strip off the end of the filename
for index,row in df.iterrows():
    #file = os.path.basename(row.to_string()).strip('.gz')
    #file = os.path.basename(df.loc[0,'filename']).strip('.gz')
    file = os.path.basename(row['filename'])
    df.loc[index,'filename'] = file
df['source'] = 'google'
df_go = df
df_go.head()
```

```python
#for the overlap I only need the filename
df = pd.read_csv(os.path.join(dataDir,'overlap.txt'),sep='\t',header=None,comment = '#',names = ['filename','cyPath','gPath'])
df = df.drop(columns = ['cyPath','gPath'])
df['source']='cyANDgoogle'
df_o = df
```

```python
result = pd.concat([df_o,df_co,df_go],ignore_index=True)
```

```python
def trimSuffix(one):
    one = one.removesuffix('.gz').removesuffix('.fastq').removesuffix('_fastqc.html').removesuffix('_fastqc.zip')      
    return one
```

```python
for index,row in result.iterrows():
    result.loc[index,'filename'] = trimSuffix(result.loc[index,'filename'])
```

```python
#result.to_csv(os.path.join(dataDir,'filelist_concatenated.csv'))
result.to_csv('../test_data/filelist_concatenated.csv')
```
