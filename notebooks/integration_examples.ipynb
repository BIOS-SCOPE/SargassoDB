# Examples of how to integrate analysis with data retrieval from FastAPI

```python
# setup
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import requests
%matplotlib inline
```

## Retrieving casts by depth

```python
url = "http://127.0.0.1:8000/niskins/"
params = {"depth_1": 0, "depth_2": 40}
response = requests.get(url, params=params)

if response.status_code == 200:
    data = response.json()
    df = pd.DataFrame(data)
    print(df)
else:
    print(f"Request failed with status code {response.status_code}")
```
