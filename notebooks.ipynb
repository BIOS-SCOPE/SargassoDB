```python
df = pd.read_csv('only_in_cyverse.txt',sep='\t',header=None,comment = '#',names = ['filename'])

#strip off the end of the filename
for index,row in df.iterrows():
    #file = os.path.basename(row.to_string()).strip('.gz')
    file = os.path.basename(row['filename'])
    df.loc[index,'filename'] = file

df['source'] = 'cyverse'
df_co = df
```

```python
df = pd.read_csv('only_in_gdrive.txt',sep='\t',header=None,comment = '#',names = ['filename'])
#strip off the end of the filename
for index,row in df.iterrows():
    #file = os.path.basename(row.to_string()).strip('.gz')
    #file = os.path.basename(df.loc[0,'filename']).strip('.gz')
    file = os.path.basename(row['filename'])
    df.loc[index,'filename'] = file
df['source'] = 'google'
df_go = df
df_go.head()
```

```python
#for the overlap I only need the filename
df = pd.read_csv('overlap.txt',sep='\t',header=None,comment = '#',names = ['filename','cyPath','gPath'])
df = df.drop(columns = ['cyPath','gPath'])
df['source']='cyANDgoogle'
df_o = df
```

```python
result = pd.concat([df_o,df_co,df_go])
```
